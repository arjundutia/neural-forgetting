# Hebbian Learning

## Question

How can repeated co-activation of neurons strengthen the connection between them over time?

## What is Hebbian Learning?

Hebbian learning is a theory of synaptic plasticity in which the connection between two neurons strengthens when the neurons are repeatedly active together.

If the activation of one neuron repeatedly contributes to the activation of another neuron, the synaptic connection between them becomes stronger. This allows patterns of neural activity that frequently occur together to become associated with one another.

This idea was proposed by Donald Hebb in 1949 and is commonly summarized as:

> "Neurons that fire together, wire together."

Hebbian learning provides a simple mechanism through which experience can modify neural connections, allowing learning and memory to emerge from changes in synaptic strength.

## Hebbian Learning Rule

The standard simplified formula for the Hebbian learning rule is:

w_final = w_initial + ηxy

where:

- (w) is the connection strength between two neurons
- (η) is the learning rate
- (x) is the activity of the presynaptic/sending neuron (measured from 0 to 1)
- (y) is the activity of the postsynaptic/receiving neuron (measured from 0 to 1)

## Experiment 1: Binary Neural Activity

```python
import numpy as np
import matplotlib.pyplot as plt

for presynaptic_activity in range(0,2):
    for postsynaptic_activity in range(0,2):

        connection_strength_initial = 0.1
        learning_rate = 0.05

        connection_strengths = [connection_strength_initial]
        events = [0]

        for i in range(1, 21):
            connection_strength_final = connection_strength_initial + learning_rate * presynaptic_activity * postsynaptic_activity
            connection_strength_initial = connection_strength_final
            connection_strengths.append(connection_strength_initial)
            events.append(i)

        label = f"{presynaptic_activity}, {postsynaptic_activity}"
        plt.plot(events, connection_strengths, label=label)

        
plt.xlabel("Event")
plt.ylabel("Connection Strength")
plt.title("Hebbian Learning Plot")
plt.legend()
plt.show()
```
